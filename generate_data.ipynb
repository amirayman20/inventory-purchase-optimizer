```python
import pandas as pd
import numpy as np
import os

np.random.seed(42)

# ============================================================
# SETTINGS
# ============================================================

N_ITEMS    = 30
N_BRANCHES = 25

# ============================================================
# ITEMS (ITEM_CODE = رقم بدل SKU-001)
# ============================================================

items = [(i, f"Product {str(i).zfill(3)}") for i in range(1, N_ITEMS + 1)]

# ============================================================
# BRANCHES — توزيع واقعي: فروع قوية قليلة، ضعيفة كتير
# ============================================================

branches = [f"BR-{str(i).zfill(2)}" for i in range(1, N_BRANCHES + 1)]

# Sales weights: أول 3 فروع قوية، وسط متوسط، باقي ضعيف
raw_weights = (
    [0.12, 0.10, 0.09] +
    [0.05, 0.04, 0.04, 0.03, 0.03] +
    [0.02] * 5 +
    [0.01] * (N_BRANCHES - 13)
)
raw_weights = raw_weights[:N_BRANCHES]
total = sum(raw_weights)
branch_weights = [w / total for w in raw_weights]

# ============================================================
# WAREHOUSE — ارصدة.xlsx
# ============================================================

warehouse_rows = []
for code, name in items:
    warehouse_rows.append({
        "ITEM_CODE":   code,
        "NAME_E":      name,
        "BAL":         np.random.randint(20, 300),
    })

warehouse_df = pd.DataFrame(warehouse_rows)

# ============================================================
# BRANCHES — مبيعات.xlsx
# ============================================================

branch_rows = []
for code, name in items:
    base_sales = np.random.randint(10, 200)
    unit_cost  = round(np.random.uniform(5, 150), 2)
    sale_price = round(unit_cost * np.random.uniform(1.15, 1.40), 2)

    for i, branch in enumerate(branches):
        sales = max(0, int(base_sales * branch_weights[i] * np.random.uniform(0.5, 1.5)))

        # فروع ضعيفة عندها رصيد أعلى نسبياً من مبيعاتها
        if i >= 13:
            bal = np.random.randint(5, 40)
        elif i >= 5:
            bal = np.random.randint(3, 20)
        else:
            bal = np.random.randint(1, 10)

        branch_rows.append({
            "ITEM_CODE":   code,
            "ITEM_NAME":   name,
            "STORE_NAME":  branch,
            "SALES_QTY":   sales,
            "BALANCE":     bal,
            "UNIT_COST":   unit_cost,
            "SALES_PRICE": sale_price,
        })

branches_df = pd.DataFrame(branch_rows)

# ============================================================
# SAVE
# ============================================================

user_home = os.path.expanduser("~")
for path in [
    os.path.join(user_home, "Desktop"),
    os.path.join(user_home, "سطح المكتب"),
    os.path.join(user_home, "OneDrive", "Desktop"),
    os.path.join(user_home, "OneDrive", "سطح المكتب"),
]:
    if os.path.exists(path):
        desktop = path
        break

warehouse_df.to_excel(os.path.join(desktop, "ارصدة.xlsx"),  index=False)
branches_df.to_excel(os.path.join(desktop,  "مبيعات.xlsx"), index=False)

print("✔️ تم إنشاء الملفات")
print(f"   ارصدة.xlsx  : {len(warehouse_df)} items")
print(f"   مبيعات.xlsx : {len(branches_df)} rows ({N_ITEMS} items × {N_BRANCHES} branches)")
```
